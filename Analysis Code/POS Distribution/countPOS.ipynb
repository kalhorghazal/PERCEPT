```python
import pandas as pd
import numpy as np
import ast
from collections import Counter
from itertools import chain
```

```python
df = pd.read_excel("PERCEPT.xlsx")
df.head()
```

```python
df['POS'] = df['POS'].apply(ast.literal_eval)
```

```python
allPOS = []

for index, row in df.iterrows():
  allPOS.extend(row["POS"])
```

```python
freq = Counter(chain.from_iterable(df["POS"]))

for item, count in freq.items():
  print(item,count, "***", count/len(allPOS))
```

```python
posPerPlatform = (
    df.groupby("Platform")["POS"]
      .apply(lambda col: Counter(chain.from_iterable(col)))
)
```
