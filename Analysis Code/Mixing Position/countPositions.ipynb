```python
import pandas as pd
import numpy as np
import ast
from collections import Counter
```

```python
df = pd.read_excel("PERCEPT.xlsx")
df.head()
```

```python
df['Positions'] = df['Positions'].apply(ast.literal_eval)
```

```python
allPositions = []

for index, row in df.iterrows():
  allPositions.extend(row["Positions"])
```

```python
print(len(allPositions))
```

```python
freq = Counter(allPositions)
for item, count in freq.items():
  print(item, count)
```

```python
df_exp = df.explode("Positions")
df_exp = df_exp[df_exp["Positions"].notna()]
```

```python
counts = (
    df_exp
    .groupby(["Platform", "Positions"])
    .size()
    .reset_index(name="count")
)
```
