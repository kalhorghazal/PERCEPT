```python
import pandas as pd
import numpy as np
import ast
from collections import Counter
```

```python
df = pd.read_excel("PERCEPT.xlsx")
df.head()
```

```python
def getPosition(text, cm_words):
    L = len(text)
    working_text = text
    positions = []

    for cm in cm_words:
        idx = working_text.find(cm)
        if idx == -1:
            positions.append(None)
            continue

        start_char = len(text) - len(working_text) + idx
        p = start_char / (L - 1) if L > 1 else 0.0

        if p <= 0.33:
            label = "initial"
        elif p <= 0.66:
            label = "medial"
        else:
            label = "final"

        positions.append(label)

        # mask matched span
        working_text = (
            working_text[:idx] + " " * len(cm) + working_text[idx + len(cm):]
        )

    return positions   # ✅ always a list
```

```python
df["Positions"] = None
```

```python
for index, row in df.iterrows():
    result = getPosition(row["Text"], ast.literal_eval(row["Words"]))
    df.at[index, "Positions"] = (
        result if isinstance(result, list) else [result]
    )
```
